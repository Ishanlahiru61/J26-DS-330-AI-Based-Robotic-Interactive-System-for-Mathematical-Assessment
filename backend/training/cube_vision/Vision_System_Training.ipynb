### Persistent Model Comparison (Runs even after a Runtime Timeout/Reset)

Since your results are saved on **Google Drive**, you can execute this cell at any point (even if your runtime restarts) to load the metrics and find the best model.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
from google.colab import drive

# Ensure drive is mounted
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

# Map our model names to their persistent results in Google Drive
# Note: YOLO11n was saved locally, so we also check if we can copy or find it.
run_paths = {
    'YOLO11n': '/content/runs/detect/train/results.csv', # If runtime is active
    'YOLO11s': '/content/drive/MyDrive/YOLO_Results/model_2_yolo11s/results.csv',
    'YOLOv8n': '/content/drive/MyDrive/YOLO_Results/model_3_yolov8n/results.csv',
    'YOLOv10n': '/content/drive/MyDrive/YOLO_Results/model_4_yolov10n-3/results.csv',
    'YOLO26n': '/content/drive/MyDrive/YOLO_Results/model_5_yolo26n/results.csv'
}

summary_list = []

for name, path in run_paths.items():
    if os.path.exists(path):
        df = pd.read_csv(path)
        df.columns = df.columns.str.strip()

        # Locate metrics from the row with the best mAP50-95 score
        best_idx = df['metrics/mAP50-95(B)'].idxmax()
        best_row = df.loc[best_idx]

        summary_list.append({
            'Model': name,
            'Best Epoch': int(best_row['epoch']),
            'mAP50 (B)': best_row['metrics/mAP50(B)'],
            'mAP50-95 (B)': best_row['metrics/mAP50-95(B)'],
            'Precision (B)': best_row['metrics/precision(B)'],
            'Recall (B)': best_row['metrics/recall(B)']
        })
    else:
        print(f"Skipping {name}: Results file not found yet (model may still be training or file is in another path).")

if summary_list:
    comparison_df = pd.DataFrame(summary_list)
    print("\n--- Model Comparison Table (Sorted by mAP50-95) ---")
    display(comparison_df.sort_values(by='mAP50-95 (B)', ascending=False))

    # Plotting comparison graphs
    fig, ax = plt.subplots(1, 2, figsize=(15, 6))

    comparison_df.plot(x='Model', y=['mAP50 (B)', 'mAP50-95 (B)'], kind='bar', ax=ax[0])
    ax[0].set_title('Precision Metrics (mAP50 vs mAP50-95)')
    ax[0].set_ylabel('Score')
    ax[0].set_ylim(0.7, 1.02)
    ax[0].grid(axis='y', linestyle='--', alpha=0.5)

    comparison_df.plot(x='Model', y=['Precision (B)', 'Recall (B)'], kind='bar', ax=ax[1])
    ax[1].set_title('Detection Metrics (Precision vs Recall)')
    ax[1].set_ylabel('Score')
    ax[1].set_ylim(0.7, 1.02)
    ax[1].grid(axis='y', linestyle='--', alpha=0.5)

    plt.tight_layout()
    plt.show()
else:
    print("No results were found on Google Drive. Make sure training has started and created 'results.csv' files.")
```

### Training Loss Curves Comparison
This cell loads the training history from each model's `results.csv` stored on Google Drive and plots the Box Loss and Class Loss over the epochs.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Configure paths pointing to results on Google Drive
run_paths = {
    'YOLO11n': '/content/drive/MyDrive/YOLO_Results/model_1_yolo11n/results.csv',
    'YOLO11s': '/content/drive/MyDrive/YOLO_Results/model_2_yolo11s/results.csv',
    'YOLOv8n': '/content/drive/MyDrive/YOLO_Results/model_3_yolov8n/results.csv',
    'YOLOv10n': '/content/drive/MyDrive/YOLO_Results/model_4_yolov10n-3/results.csv',
    'YOLO26n': '/content/drive/MyDrive/YOLO_Results/model_5_yolo26n/results.csv'
}

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

for name, path in run_paths.items():
    if os.path.exists(path):
        df = pd.read_csv(path)
        df.columns = df.columns.str.strip()

        # Skip mock results or short/invalid data (like our 1-row reconstructed YOLO11n if it has no epoch history)
        if len(df) <= 1:
            print(f"Note: {name} only has final validation metrics or single epoch data, skipping loss curve.")
            continue

        # Plot Box Loss
        if 'train/box_loss' in df.columns:
            axes[0].plot(df['epoch'], df['train/box_loss'], label=name, alpha=0.8)

        # Plot Class Loss
        if 'train/cls_loss' in df.columns:
            axes[1].plot(df['epoch'], df['train/cls_loss'], label=name, alpha=0.8)
    else:
        print(f"Skipping {name}: Results file not found.")

# Formatting Box Loss plot
axes[0].set_title('Training Box Loss Curve')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Box Loss')
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend()

# Formatting Class Loss plot
axes[1].set_title('Training Class Loss Curve')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Class Loss')
axes[1].grid(True, linestyle='--', alpha=0.5)
axes[1].legend()

plt.tight_layout()
plt.show()
```
