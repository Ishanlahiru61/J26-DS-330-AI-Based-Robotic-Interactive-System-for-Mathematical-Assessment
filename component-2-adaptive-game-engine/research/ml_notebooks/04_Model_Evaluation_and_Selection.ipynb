# Model Evaluation and Selection

Loads frozen result tables and the separately derived threshold-0.5 diagnostics. It distinguishes original probability metrics from post-freeze classification metrics.

**Safety:** default execution is lightweight. Expensive full-data work runs only behind an explicit environment flag.

```python
from pathlib import Path
import os
import json
import pandas as pd

WORKSPACE = Path(
    r"C:\Users\Monil A\Desktop\Research Documents\ASSISTments Dataset"
)

if not WORKSPACE.is_dir():
    raise FileNotFoundError(
        f"ASSISTments research workspace not found: {WORKSPACE}"
    )

print("Using workspace:", WORKSPACE)
```

```python
step3_out = WORKSPACE / '03_Experimental_Modeling' / 'outputs'
final = WORKSPACE.parent / '08_C2_ML_PP1_Finalization'
results = pd.read_csv(step3_out / 'ASSISTments_Step3_Model_Results.csv')
native_test = results[(results['split']=='test') & (results['calibration']=='native')]
display(native_test[['experiment_id','Log Loss','Brier Score','ROC-AUC','PR-AUC','ECE_10_equal_frequency']])
```

```python
selected = results[(results.experiment_id=='H2_GB_RT_OFF') & (results.split=='test') & (results.calibration=='platt_retained')]
display(selected)
assert round(float(selected['Log Loss'].iloc[0]),6) == 0.534442
assert round(float(selected['ROC-AUC'].iloc[0]),6) == 0.750158
print('Frozen selected metrics verified: PASS')
```
