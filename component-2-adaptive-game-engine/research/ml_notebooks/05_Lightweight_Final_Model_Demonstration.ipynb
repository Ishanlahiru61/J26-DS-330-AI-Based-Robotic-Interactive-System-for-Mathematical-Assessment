# Lightweight Final Model Demonstration

Verifies hashes, loads the frozen H2_GB_RT_OFF_v1 model and validation-fitted Platt calibrator, and reproduces one saved ASSISTments test-row probability without retraining.

**Safety:** default execution is lightweight. Expensive full-data work runs only behind an explicit environment flag.

```python
from pathlib import Path
import os, json
import pandas as pd

candidate = Path(os.environ.get('ASSISTMENTS_WORKSPACE', Path.cwd().parents[1] / 'ASSISTments Dataset')).resolve()
if not candidate.exists():
    raise FileNotFoundError('Set ASSISTMENTS_WORKSPACE to the local ASSISTments Dataset folder')
WORKSPACE = candidate
print('Using workspace:', WORKSPACE)
```

```python
import subprocess, sys
demo = WORKSPACE / '06_PP1_Final_Component2' / 'component-2-adaptive-game-engine' / 'research' / 'assistments' / 'public_model_demo.py'
completed = subprocess.run([sys.executable, str(demo), '--workspace', str(WORKSPACE)], capture_output=True, text=True)
print(completed.stdout)
if completed.returncode != 0:
    raise RuntimeError(completed.stderr or completed.stdout)
result = json.loads(completed.stdout)
assert result['status'] == 'PASS'
print('Fresh frozen-model inference verification: PASS')
```

## Interpretation boundary

The output is a conditional first-attempt correctness probability for an ASSISTments observation. It is **not** a local robot-game score, dyscalculia probability, diagnosis, mastery label, or causal estimate of what would happen under a different activity. The operational prototype correctly keeps local `predicted_probability` null until a validated local feature adapter and transfer study exist.
